```python

# Nearest Neighbour Algorithm for TSP

# Distance matrix
distance = [
    [0, 10, 15, 20, 25],
    [10, 0, 35, 25, 30],
    [15, 35, 0, 30, 20],
    [20, 25, 30, 0, 15],
    [25, 30, 20, 15, 0]
]

cities = ["A", "B", "C", "D", "E"]

n = len(distance)

# Initially, no city has been visited
visited = [False] * n

# Start from city A (index 0)
tour = [0]
current = 0
visited[current] = True

# Visit all remaining cities
for step in range(n - 1):

    nearest = -1
    minimum_distance = float('inf')

    # Find the nearest unvisited city
    for city in range(n):

        if not visited[city]:

            if distance[current][city] < minimum_distance:
                minimum_distance = distance[current][city]
                nearest = city

    # Move to the nearest city
    current = nearest

    # Mark it as visited
    visited[current] = True

    # Add it to the tour
    tour.append(current)

# Return to the starting city
tour.append(0)

# Calculate total distance
total_distance = 0

for i in range(len(tour) - 1):
    total_distance += distance[tour[i]][tour[i + 1]]

# Display results
print("Tour:", end=" ")

for city in tour:
    print(cities[city], end=" ")

print("\nTotal distance:", total_distance)
```
