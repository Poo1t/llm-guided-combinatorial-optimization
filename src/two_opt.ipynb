```python

# 2-opt Algorithm for TSP

distance = [
    [0, 10, 15, 20, 25],
    [10, 0, 35, 25, 30],
    [15, 35, 0, 30, 20],
    [20, 25, 30, 0, 15],
    [25, 30, 20, 15, 0]
]

cities = ["A", "B", "C", "D", "E"]

# Calculate total tour distance
def calculate_distance(tour):
    total = 0

    for i in range(len(tour) - 1):
        total += distance[tour[i]][tour[i + 1]]

    return total


# Apply 2-opt
def two_opt(tour):

    best_tour = tour[:]
    best_distance = calculate_distance(best_tour)

    improved = True

    while improved:

        improved = False

        # Select first edge position
        for i in range(1, len(best_tour) - 2):

            # Select second edge position
            for j in range(i + 1, len(best_tour) - 1):

                # Reverse the section between i and j
                new_tour = (
                    best_tour[:i]
                    + best_tour[i:j + 1][::-1]
                    + best_tour[j + 1:]
                )

                new_distance = calculate_distance(new_tour)

                # Accept if the new tour is shorter
                if new_distance < best_distance:

                    best_tour = new_tour
                    best_distance = new_distance
                    improved = True

                    break

            if improved:
                break

    return best_tour, best_distance


# Initial tour from Nearest Neighbour
initial_tour = [0, 1, 3, 4, 2, 0]

print("Initial tour:", [cities[i] for i in initial_tour])
print("Initial distance:", calculate_distance(initial_tour))

# Optimize using 2-opt
optimized_tour, optimized_distance = two_opt(initial_tour)

print("\nOptimized tour:", [cities[i] for i in optimized_tour])
print("Optimized distance:", optimized_distance)
```
